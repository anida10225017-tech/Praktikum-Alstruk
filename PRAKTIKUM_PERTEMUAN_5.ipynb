```python
# Praktikum 1 Mengenali Tipe Data
# Membuat variabel
nama = "Fajar"
umur = 19
berat = 54.5
terdaftar = True

# Menampilkan nilai dan tipe data
print(nama)
print(type(nama))

print(umur)
print(type(umur))

print(berat)
print(type(berat))

print(terdaftar)
print(type(terdaftar))

# Mengubah nilai umur menjadi 20
umur = 20
print(umur)
print(type(umur))
```

```python
# Praktikum 2 Oprator Aritmatika

harga = 12500
jumlah = 4

# Menghitung total harga
total_harga = harga*jumlah

# Menghitung sisa pembagian dengan 2.000
sisa = total_harga % 2000

# Menghitung harga setelah diskon Rp.5000
harga_diskon = total_harga-5000

# Menampilkan hasil
print("Harga per buku : Rp.",harga)
print("Jumlah buku :", jumlah)
print("Total harga : Rp", total_harga)
print("Sisa pembagian total harga dengan 2.000: Rp", sisa)
print("Harga setelah dikson : Rp", harga_diskon)
```

```python
# Praktikum 3 Operator Perbandingan dan Logika
# Buat variabel nilai dan kehadiran
nilai = 82
kehadiran = 90

# Gunakan operator >= untuk memeriksa masing masing syarat
syarat_nilai = nilai >= 75
syarat_kehadiran = kehadiran >= 80

print("Apakah nilai memenuhi syarat? (>= 75) :", syarat_nilai)
print("Apakah kehadiran memenuhi syarat? (>= 80 %):", syarat_kehadiran)

# Gunakan operator and untuk menentukan kelulusan
Lulus = (nilai >= 75) and (kehadiran >= 80)
print("Status kelulusan (and)", Lulus)

# Ubah nilai menjadi 70
nilai = 70
Lulus_baru = (nilai >= 75) and (kehadiran >= 80)
print("\nJika nilai diubah menjadi 70:")
print("Status kelulusan baru(and):", Lulus_baru)

# Coba gunakan or
Lulus_or = (nilai >= 75) or (kehadiran >= 80)
print("Status dengan operator or :", Lulus_or)
```

```python
# Praktikum 4 Operator Penugasan
# Simpan poin awal dalam variabel poin
poin = 100

print("Nilai awal :", poin)
poin += 25 # Tambah 25
print("Setelah += 25:", poin)
poin -= 10 # kurang 10
print("Setelah -= 10:", poin)
poin *= 2 # kali 2
print("Setelah *2:", poin)
poin/=2 # bagi 2
print("Setelah /=2:", poin)
```

```python
# Praktikum 5 Struktur Data
nilai = [78, 85, 90, 85]
mahasiswa = {"nama" : "Nadia", "kelas": "IF-A"}
koordinat = (5, 8)
angka = {2, 4, 4, 6, 8}

# Tampilkan nilai pertama dari list
print("Nilai pertama list:", nilai[0])

# Tambahkan nilai 95 ke list dan tampilkan list terbaru
nilai.append(95)
print("list terbaru:", nilai)

# Tampilkan koordinat pertama dari tauple
print("koordinat pertama:", koordinat[0])

# Tampilkan angka unik dengan sorted(angka)
print("Angka unik terurut:", sorted(angka))
```

```python
# Praktikum 5 Tantangan Terintegrasi - Program Rekap Nilai
# Data Mahasiswa
nama = "Anida Wardatun Nadiyah"
nilai_tugas = 90
nilai_ujian = 92  # Perhatikan pada gambar pertama ada sedikit ralat variabel di baris ini menjadi nilai_ujian (menggunakan underscore) agar tidak error

# Menghitung rata-rata
rata_rata = (nilai_tugas + nilai_ujian) / 2

# Memeriksa status kelulusan
lulus = rata_rata >= 75

# Menampilkan hasil
print("=== PROGRAM REKAP NILAI ===")
print("Nama mahasiswa:", nama)
print("Nilai tugas:", nilai_tugas)
print("Nilai ujian:", nilai_ujian)
print("Rata-rata:", rata_rata)
print("Status lulus:", lulus)

# Pengujian kedua dengan nilai berbeda
print("\n=== PENGUJIAN KEDUA ===")

nama = "Anida Wardatun Nadiyah"
nilai_tugas = 85
nilai_ujian = 77

rata_rata = (nilai_tugas + nilai_ujian) / 2
lulus = rata_rata >= 75

print("Nama mahasiswa:", nama)
print("Nilai tugas:", nilai_tugas)
print("Nilai ujian:", nilai_ujian)
print("Rata-rata:", rata_rata)
print("Status lulus:", lulus)


```
